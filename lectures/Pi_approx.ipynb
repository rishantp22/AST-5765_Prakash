```python
import math
def apprx_pi(N):

    if N < 0: # N has to be non-negative
        raise ValueError("N should be non-negative.")

    N = math.floor(N) # using this function to round off to nearest integer

    t = 0.0 #initial value

    # Sum from k = 0 to k = N
    for k in range(N + 1):
        term = (-3) ** (-k) / (2 * k + 1)
        t += term
        
    pi_approx = math.sqrt(12) * t

    print(pi_approx)
    return pi_approx

apprx_pi(22)
```
